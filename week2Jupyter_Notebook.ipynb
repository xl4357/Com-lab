```python
spam = [2, 4, 6, 8, 10]
spam[2] = 'hello'
print(spam)

letters = ['a', 'b', 'c', 'd']

print(letters[int(int('3'*2)//11)])

print(letters[-1])

print(letters[:2])

my_list = [1, 2, 3]
my_tuple = tuple(my_list)
print(my_tuple)

another_tuple = (4, 5, 6)
another_list = list(another_tuple)
print(another_list)

def comma_code(items):
    if len(items) == 0:
        return ''
    elif len(items) == 1:
        return str(items[0])
    else:
        return ', '.join(str(i) for i in items[:-1]) + ', and ' + str(items[-1])

spam = ['apples', 'bananas', 'tofu', 'cats']
print(comma_code(spam))
```
